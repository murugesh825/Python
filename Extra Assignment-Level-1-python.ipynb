```python
# Addition, Subtraction, Multiplication, Float Division, 
# Floor Division, Module, Power
num1=int(input('Enter Num1 : '))
num2=int(input('Enter Num2 : '))
print('----------------------------------------------------------')
print('Addition of given numbers       : ',num1+num2)
print('Subtraction of given numbers    : ',num1-num2)
print('Multiplication of given numbers : ',num1*num2)
print('Float Division of given numbers : ',num1/num2)
print('Floor Division of given numbers : ',num1//num2)
print('Modulo of given numbers         : ',num1%num2)
print('Power of given numbers          : ',num1**num2)
print('----------------------------------------------------------')
```
